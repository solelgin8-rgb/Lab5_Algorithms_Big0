```python
# Student-Specific Inputs
LAST_NAME = "Soliva"  
STUDENT_ID = "TUPM-26-4154"  
FAVORITE_ARTIST = "Taylor Swift"  

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(c.upper()) - 64 for c in FAVORITE_ARTIST if c.isalpha())

print("Student Inputs Configured Successfully!")
print(f"SEED_DIGIT: {SEED_DIGIT} | ID_SUM: {ID_SUM} | ARTIST_SUM: {ARTIST_SUM}")
```

```python
# Exercise 1: Emergency Dispatch Record Search
# 1. Generate original sorted incident records
original_records = sorted([
    (ID_SUM * i + SEED_DIGIT) % 1000 
    for i in range(1, 15)
])

# 2. Determine rotation point using SEED_DIGIT and rotate list[cite: 14]
rotation_point = SEED_DIGIT % len(original_records)
rotated_records = original_records[rotation_point:] + original_records[:rotation_point]

# 3. Generate target incident using ID_SUM and ARTIST_SUM[cite: 14]
target_incident = ((ID_SUM + ARTIST_SUM) % 1000)

def rotated_binary_search(data, target):
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        
        if data[low] <= data[mid]:
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons

def linear_search_dispatch(data, target):
    comparisons = 0
    for i, val in enumerate(data):
        comparisons += 1
        if val == target:
            return i, comparisons
    return -1, comparisons

index_rot, comp_rot = rotated_binary_search(rotated_records, target_incident)
index_lin, comp_lin = linear_search_dispatch(rotated_records, target_incident)

print("--- Exercise 1: Emergency Dispatch Record Search ---")
print("Original Dataset:", original_records)
print("Rotated Dataset: ", rotated_records)
print("Target Incident:", target_incident)
print(f"Rotated Search -> Position: {index_rot}, Comparisons: {comp_rot}")
print(f"Linear Search  -> Position: {index_lin}, Comparisons: {comp_lin}")
```

```python
# Exercise 2: University Enrollment Records
# 1. Generate two sorted datasets using student inputs
dataset_a = sorted([
    (SEED_DIGIT * i + ID_SUM) % 200 
    for i in range(1, 8)
])

dataset_b = sorted([
    (ARTIST_LENGTH * i + NAME_LENGTH) % 200 
    for i in range(1, 8)
])

def merge_datasets(a, b):
    consolidated = []
    i = j = comparisons = 0
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            consolidated.append(a[i])
            i += 1
        else:
            consolidated.append(b[j])
            j += 1
    while i < len(a):
        consolidated.append(a[i])
        i += 1
    while j < len(b):
        consolidated.append(b[j])
        j += 1
    return consolidated, comparisons

# 2. Consolidate into one ordered collection[cite: 15]
consolidated_data, merge_comparisons = merge_datasets(dataset_a, dataset_b)

print("\n--- Exercise 2: University Enrollment Records ---")
print("Department A Dataset:", dataset_a)
print("Department B Dataset:", dataset_b)
print("Consolidated Dataset:", consolidated_data)
print("Records Before Consolidation:", len(dataset_a) + len(dataset_b))
print("Records After Consolidation: ", len(consolidated_data))
print("Total Comparisons Performed:", merge_comparisons)
print("Final Order Verified:        ", consolidated_data == sorted(consolidated_data))
```

```python
# Exercise 3: E-Commerce Order Processing
import random

# 1. Generate base order dataset and input conditions
base_orders = [
    (ID_SUM * i + SEED_DIGIT * 7) % 500 
    for i in range(1, 20)
]

input_conditions = {
    "Sorted": sorted(base_orders),
    "Reverse-Ordered": sorted(base_orders, reverse=True),
    "Random": random.sample(base_orders, len(base_orders))
}

def bubble_sort_metrics(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return comparisons, swaps

def selection_sort_metrics(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return comparisons, swaps

def insertion_sort_metrics(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return comparisons, shifts

print("\n--- Exercise 3: E-Commerce Order Processing Performance ---")
for condition, dataset in input_conditions.items():
    b_comp, b_swap = bubble_sort_metrics(dataset)
    s_comp, s_swap = selection_sort_metrics(dataset)
    i_comp, i_shift = insertion_sort_metrics(dataset)
    
    print(f"\nCondition: {condition}")
    print(f"  Bubble Sort    -> Comparisons: {b_comp}, Movements (Swaps): {b_swap}")
    print(f"  Selection Sort -> Comparisons: {s_comp}, Movements (Swaps): {s_swap}")
    print(f"  Insertion Sort -> Comparisons: {i_comp}, Movements (Shifts): {i_shift}")
```
